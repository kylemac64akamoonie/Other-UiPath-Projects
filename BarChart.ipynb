```python
# prompt: 1.	Ask each of your team mates what their favourite programming language is. Create a bar chart showing the popularity of each language. (make up the data)

import matplotlib.pyplot as plt

# Sample data (replace with your team's actual responses)
languages = ['Python', 'Java', 'JavaScript', 'C++', 'Go', 'R']
votes = [5, 3, 2, 4, 1, 2]

# Create the bar chart
plt.figure(figsize=(10, 6))
plt.bar(languages, votes, color='skyblue')
plt.xlabel("Programming Languages")
plt.ylabel("Number of Votes")
plt.title("Team's Favorite Programming Languages")
plt.grid(axis='y', alpha=0.75)
plt.show()
```
